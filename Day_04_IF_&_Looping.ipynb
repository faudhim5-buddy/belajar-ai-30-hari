```python
nilai = 15

if nilai >= 90:
  print("A - MUMTAZ")
elif nilai >= 80:
  print("B - JAYYID JIDDAN")
elif nilai >= 65:
  print("C - JAYYID")
elif nilai >= 50:
  print("D - MAQBUL")
elif nilai >= 25:
  print("E - DHOIF")
else:
  print("BELAJAR BLOG! NILAI LU DHOIF JIDDAN!")
```

```python
buah = ["anggur", "semangka", "melon", "pisang", "buah kaktus"]

for item in buah:
  print("aku suka", item)
```

```python
angka = [9, 4, 20, 13, 5, 37, 8, 66]

for nilai in angka:
  if nilai > 10:
    print(nilai, "lebih besar dari 10" )
  elif nilai < 10:
    print(nilai, "kurang dari 10")
  else:
    print(nilai, "sama dengan 10")
```

```python
hitung = 5

while hitung <= 10:
  print("hitungan ke", hitung)
  hitung = hitung + 1

print("Selesai!")
```
