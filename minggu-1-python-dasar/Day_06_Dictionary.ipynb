```python
mahasiswa = {
    "nama" : "Fauzan",
    "umur" : 25,
    "jurusan" : "Hukum Islam",
    "beasiswa" : False,
    "semester" : "belum diisi",
    }

print(mahasiswa)
print(mahasiswa["nama"])
print(mahasiswa["jurusan"])
```

```python
mahasiswa["sememster"] = "terakhir"
mahasiswa["hobi"] = "mendaki"

print(mahasiswa)
```

```python
nilai_ujian = {
    "sastra arab" : 95,
    "bahasa inggris" : 100,
    "tafsir" : 85,
}

for mapel in nilai_ujian:
  print(mapel, ":", nilai_ujian[mapel])
```
