```python
import numpy as np
from sklearn.model_selection import train_test_split

data = np.array([
    [160, 50], [165, 55], [170, 60], [175, 65], [180, 70],
    [155, 70], [160, 75], [165, 80], [170, 85], [175, 90],
    [162, 52], [167, 57], [172, 62], [177, 67], [182, 72],
    [157, 72], [162, 77], [167, 82], [172, 87], [177, 92],
])

label = [
    "ideal", "ideal", "ideal", "ideal", "ideal",
    "overweight", "overweight", "overweight", "overweight", "overweight",
    "ideal", "ideal", "ideal", "ideal", "ideal",
    "overweight", "overweight", "overweight", "overweight", "overweight",
]

X_latih, X_uji, y_latih, y_uji = train_test_split(
    data, label, test_size=0.2, random_state=32
)

print("Jumlah data keseluruhan:", len(data))
print("Jumlah data latih:", len(X_latih))
print("Jumlah data uji:", len(X_uji))
```

```python
from sklearn.neighbors import KNeighborsClassifier

model = KNeighborsClassifier(n_neighbors=3)
model.fit(X_latih, y_latih)

print("Model telah dilatih dengan", len(X_latih), "data!")
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, classification_report
from sklearn.model_selection import train_test_split
import numpy as np

data = np.array([
    [160, 50], [165, 55], [170, 60], [175, 65], [180, 70],
    [155, 70], [160, 75], [165, 80], [170, 85], [175, 90],
    [163, 71], [168,61],
    [162, 52], [167, 57], [172, 62], [177, 67], [182, 72],
    [157, 72], [162, 77], [167, 82], [172, 87], [177, 92],

])

label = [
    "ideal", "ideal", "ideal", "ideal", "ideal",
    "overweight", "overweight", "overweight", "overweight", "overweight",
    "overweight", "ideal",
    "ideal", "ideal", "ideal", "ideal", "ideal",
    "overweight", "overweight", "overweight", "overweight", "overweight",

]

X_latih, X_uji, y_latih, y_uji = train_test_split(
    data, label, test_size=0.3, random_state=42
)

model = KNeighborsClassifier(n_neighbors=5)
model.fit(X_latih, y_latih)

prediksi_latih = model.predict(X_latih)
akurasi_latih = accuracy_score(y_latih, prediksi_latih)

prediksi_uji = model.predict(X_uji)
akurasi_uji = accuracy_score(y_uji, prediksi_uji)

print("Jumlah data keseluruhan:", len(data))
print("Jumlah data latih:", len(X_latih))
print("Jumlah data uji:", len(X_uji))
print()

print("Akurasi data latih:", akurasi_latih*100, "%")
print("Akurasi data uji:", akurasi_uji*100, "%")
print()
print("Laporan detail data uji")
print(classification_report(y_uji, prediksi_uji))
```

```python
import matplotlib.pyplot as plt

kategori = ["Data Latih", "Data Uji"]
akurasi = [akurasi_latih*100, akurasi_uji*100]

plt.bar(kategori, akurasi, color=["blue", "green"])
plt.title("Perbandingan Akurasi Data Latih vs Data Uji")
plt.ylabel("Akurasi %")
plt.ylim(0, 110)

for i, v in enumerate(akurasi):
  plt.text(i, v+1, str(round(v, 1))+"%", ha="center")
plt.show()
```
