```python
import pandas as pd
import numpy as np
from sklearn.neighbors import NearestNeighbors
from sklearn.preprocessing import MinMaxScaler

data_buku = {
    "judul" : [
        "Bumi Manusia", "Laskar Pelangi", "Negeri 5 Menara",
        "Ayat-Ayat Cinta", "Harry Potter", "The Alchemist",
        "Rich Dad Poor Dad", "Atomic Habits", "Clean Code",
        "Python Crash Course", "Overthingking is my hobby and i hate it",
        "Jika tidak pernah jadi apa-apa", "Petualangan Don Quixote",
        "Dark Psychology", "The Principles of Power",
        "Toko Tukar Tambah Nasib", "Animal Farm",
    ],
    "genre" : [
         "fiksi sejarah", "fiksi", "fiksi islami",
        "fiksi islami", "fiksi fantasi", "fiksi motivasi",
        "nonfiksi keuangan", "nonfiksi pengembangan diri",
        "nonfiksi teknologi", "nonfiksi teknologi", "nonfiksi self help",
         "nonfiksi self help", "fiksi", "nonfiksi psikologi",
         "nonfiksi psikologi", "fiksi fantasi", "fiksi politik"
    ],
    "rating": [4.8, 4.7, 4.6, 4.5, 4.9, 4.7, 4.6, 4.8, 4.5, 4.6, 4.8, 4.7, 4.0, 4.3, 4.3, 4.5, 4.9],
    "jumlah_halaman": [535, 529, 423, 411, 309, 208, 336, 320, 431, 544, 268, 236, 124, 200, 194, 340, 148],
    "tahun": [1980, 2005, 2009, 2004, 1997, 1988, 1997, 2018, 2008, 2015, 2023, 2019, 1605, 2023, 2023, 2022, 1945]
}

df = pd.DataFrame(data_buku)

genre_encoded = pd.get_dummies(df["genre"])
fitur_numerik = df[["rating", "jumlah_halaman", "tahun"]]
X = pd.concat([fitur_numerik, genre_encoded], axis =1)

scaler = MinMaxScaler()
X_scale = scaler.fit_transform(X)

model_rekomendasi = NearestNeighbors(n_neighbors=4, metric="euclidean")
model_rekomendasi.fit(X_scale)

def rekomendasikan_buku(judul_buku):
  if judul_buku not in df["judul"].values:
    print("Buku tidak ditemukan!")
    return

  idx = df[df["judul"] == judul_buku].index[0]

  jarak, indices = model_rekomendasi.kneighbors([X_scale[idx]])

  print(f"Karena kamu suka '{judul_buku}', coba baca juga:")
  print("-" *45)
  for i, ind in enumerate(indices[0][1:]):
    print(f"{i+1}. {df['judul'][ind]} ({df['genre'][ind]}) stars {df['rating'][ind]}")

rekomendasikan_buku("Laskar Pelangi")
print()
rekomendasikan_buku("Python Crash Course")
print()
rekomendasikan_buku("Animal Farm")
print()
print("Genre setelah di-encide:")
print(genre_encoded)
```
