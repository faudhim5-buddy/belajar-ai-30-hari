```python
import matplotlib.pyplot as plt

day = [1,2,3,4,5,6,7,8,9]
grade = [60,65,70,75,80,85,90,95,99]

plt.plot(day,grade)
plt.title("Perkembangan Nilai Ujian Harian")
plt.xlabel("Day")
plt.ylabel("Grade")
plt.show()
```

```python
import matplotlib.pyplot as plt

nama = ["Agus", "Bagas", "Cania", "Darto", "Eka", "Feby"]
nilai = [88, 90, 92, 78, 95, 85]

plt.bar(nama,nilai)
plt.title("Perbangingan Nilai Murid")
plt.xlabel("Nama")
plt.ylabel("Nilai")
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

nilai_kelas = np.random.randint(50,100,100)

plt.hist(nilai_kelas, bins=10)
plt.title("Distribusi Nilai 100 Siswa")
plt.xlabel("Rentang Nilai")
plt.ylabel("Jumalah Siswa")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "nama":["Andi", "Bagus", "Citra", "Dimas", "Emily", "Farhan"],
    "nilai":[90,88,83,79,90,94],
    "jurusan":["Akutansi", "Satra", "Hukum", "Teknik Sipil", "Biologi", "DKV"]
}

df = pd.DataFrame(data)
df.plot(x="nama", y="nilai", kind="bar", title="Nilai Siswa")
plt.ylabel("Nilai")
plt.show()
```
