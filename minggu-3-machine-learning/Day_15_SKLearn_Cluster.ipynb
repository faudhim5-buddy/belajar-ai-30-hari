```python
from sklearn.cluster import KMeans
import numpy as np
import matplotlib.pyplot as plt

data = np.array([
    [200, 5], [220, 6], [210, 4], [230, 7],
    [500, 15], [520, 14], [510, 16], [520, 13],
    [900, 25], [ 920, 24], [910, 26], [930, 23],
])

kmeans = KMeans(n_clusters=3, random_state=42)
kmeans.fit(data)

print("Kelompok tiap data", kmeans.labels_)

warna = ["grey", "purple", "blue"]
for i, titik in enumerate(data) :
  kelompok = kmeans.labels_[i]
  plt.scatter(titik[0], titik[1], color=warna[kelompok])

plt.title("Hasil Clustering")
plt.xlabel("Penegeluaran")
plt.ylabel("Frekuensi")
plt.show()
```
