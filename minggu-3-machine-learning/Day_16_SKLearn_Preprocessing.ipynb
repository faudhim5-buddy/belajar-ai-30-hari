```python
import numpy as np

data = np.array([
    [25, 3000000],
    [30, 5000000],
    [35, 7000000],
    [40, 6000000],
    [45, 9000000],
])

print("Data Asli")
print(data)
print()
print("Nilai minimum tiap fitur:", data.min(axis=0).tolist())
print("Nilai maksimum tiap fitur:", data.max(axis=0).tolist())
print("Selisih rentang:", (data.max(axis=0) - data.min(axis=0)).tolist())
```

```python
from sklearn.preprocessing import MinMaxScaler
import numpy as np

data = np.array([
    [25, 3000000],
    [30, 5000000],
    [35, 7000000],
    [40, 6000000],
    [45, 9000000],
])

scaler = MinMaxScaler()
data_normal = scaler.fit_transform(data)

print("Data sebelum dinormalisasi")
print(data)
print()
print("Data setelah dinormalisasi")
print(data_normal)
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import MinMaxScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
import numpy as np

data = np.array([
    [25, 3000000], [27, 3500000], [26, 2800000],
    [45, 9000000], [48, 9500000], [46, 8800000],
    [35, 5000000], [37, 5500000], [36, 4800000],
])

label = ["junior", "junior", "junior",
        "senior", "senior", "senior",
        "mid", "mid", "mid"]

X_latih, X_uji, y_latih, y_uji = train_test_split(
    data, label, test_size=0.3, random_state=42
)


model1 = KNeighborsClassifier(n_neighbors=3)
model1.fit(X_latih, y_latih)
akurasi1 = accuracy_score(y_uji, model1.predict(X_uji))

scaler = MinMaxScaler()
X_latih_normal = scaler.fit_transform(X_latih)
X_uji_normal = scaler.transform(X_uji)

model2 = KNeighborsClassifier(n_neighbors=3)
model2.fit(X_latih_normal, y_latih_normal := y_latih)
akurasi2 = accuracy_score(y_uji, model2.predict(X_uji_normal))

print("Akurasi KKN tanpa normalisasi", akurasi1 *100, "%")
print("Akurasi KKN dengan normalisasi", akurasi2 *100, "%")
```
