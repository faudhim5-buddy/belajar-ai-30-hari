```python
import pandas as pd

data = {
    "Nama" : ["Sonya", "Yeni", "Una", "Yuna", "Bian", "Jiya"],
    "Nilai" : [85,80,90,88,82,95],
    "Jurusan" : ["Tata Busana", "Akutansi", "Kedokteran", "Teknologi Musik", "Management", "Jurnalistik"]
}

df = pd.DataFrame(data)
print(df["Nama"])
print(df["Nilai"].mean())
print(df[df["Nilai"]>89])
print()

df["Status"] = df["Nilai"].apply(lambda x : "Lulus" if x >= 75 else "Tidak Lulus" )
print(df)
```
