```python
from groq import Groq
from google.colab import userdata

groq_api_key = userdata.get('GROQ_API_KEY')

client = Groq(api_key=groq_api_key)

def analisis_sentimen(ulasan):
    prompt = f"""Analisis sentimen dari ulasan produk berikut.
Jawab hanya dengan format:
Sentimen: [positif/negatif/netral]
Alasan: [satu kalimat singkat maksimum 15 kata]

Ulasan: {ulasan}"""

    response = client.chat.completions.create(
        model="llama-3.1-8b-instant",
        messages=[
            {"role": "user", "content": prompt}
        ]
    )
    return response.choices[0].message.content

ulasan_list = [
    "Produk ini luar biasa, kualitasnya jauh melebihi ekspektasi saya!",
    "Pengiriman lambat dan barang datang dalam kondisi rusak, sangat kecewa",
    "Produknya lumayan, tidak terlalu bagus tapi juga tidak buruk",
]

for ulasan in ulasan_list:
  print("Ulasan:", ulasan)
  print(analisis_sentimen(ulasan))
  print("-" *50)
```

```python
from groq import Groq
from google.colab import userdata

groq_api_key = userdata.get('GROQ_API_KEY')

client = Groq(api_key=groq_api_key)

def chatbot():
  print("Chatbot AI siap! Ketik 'keluar' untuk berhenti.")
  print("-" *40)

  riwayat = []

  while True:
    pesan_user = input("Kamu: ")
    if pesan_user.lower() == "keluar":
      print("Sampai jumpa!")
      break

      riwayat.append({
          "role": "user",
          "content": pesan_user
      })

      response = client.chat.completions.create(
        model="llama-3.1-8b-instant",
        messages=[
            {"role": "system", "content": "Kamu adalah asisten AI yang membantu dalam bahasa Indonesia."}
        ] + riwayat
      )
      jawaban = response.choices[0].message.content

      riwayat.append({
          "role": "assistant",
          "content": jawaban
      })

      print("AI:", jawaban)
      print()

chatbot()
```
