```python
import re

def bersihkan_teks(teks):
  teks = teks.lower()
  teks = re.sub(r'[^\w\s]','', teks)
  return teks

def tokenisasi(teks):
  return teks.split()

stopwords = ["yang", "dan", "di", "ke", "dari", "untuk"
            "ini", "itu", "adalah", "dengan", "tidak", "saya"]

def hapus_stopwords(tokens):
  return [kata for kata in tokens if kata not in stopwords]

kalimat = "Saya tidak suka dengan pelayanan ini!"
bersih = bersihkan_teks(kalimat)
tokens = tokenisasi(bersih)
tokens_bersih = hapus_stopwords(tokens)

print("kalimat asli:", kalimat)
print("setelah dibersihkan:", bersih)
print("seletah di tokenisasi:", tokens)
print("setelah hapus stopwords:", tokens_bersih)
```

```python
from numpy import vectorize
from sklearn.feature_extraction.text import CountVectorizer


kalimat = [
    "film ini sangat bagus dan menarik",
    "pelayanan sangat buruk dan mengecewakan",
    "produk bagus tapi pengiriman lambat",
    "sangat puas dengan produk ini",
    "tidak puas sama sekali sangat buruk",
]

vectorizer = CountVectorizer()
X = vectorizer.fit_transform(kalimat)


print("Kosakata yang ditemukan :")
print(vectorizer.get_feature_names_out())
print()
print("Matirks angka dari teks :")
print(X.toarray())
print()

```

```python
from numpy import vectorize
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

ulasan = [
    "produk ini sangat bagus dan berkualitas",
    "saya sangat puas dengan pembelian ini",
    "pengiriman cepat produk sesuai deskripsi",
    "barang bagus harga terjangkau recommended",
    "kualitas sangat baik akan beli lagi",
    "produk ini mengecewakan dan sangat buruk",
    "kualitas jelek tidak sesuai gambar",
    "pengiriman lambat barang rusak kecewa",
    "tidak recommended sama sekali buang uang",
    "pelayanan buruk tidak sesuai deskripsi",
    "barang rusak saat diterima sangat kecewa",
    "produk mengecewakan tidak sesuai ekspektasi",
    "lumayan bagus tapi pengiriman agak lama",
    "cukup puas meski ada kekurangan kecil",
    "barang tiba cepat kualitas memuaskan",
]

label = ["postif", "postif", "postif", "postif", "postif",
         "negatif", "negatif","negatif", "negatif", "negatif",
         "negatif", "negatif",
         "positif", "positif", "positif",]

X_latih, X_uji, y_latih, y_uji = train_test_split(
    ulasan, label, test_size=0.2, random_state=42
)

vectorizer = CountVectorizer()
X_latih_vec = vectorizer.fit_transform(X_latih)
X_uji_vec = vectorizer.transform(X_uji)

model = MultinomialNB()
model.fit(X_latih_vec, y_latih)

y_pred = model.predict(X_uji_vec)
akurasi = accuracy_score(y_uji, y_pred)

print("Akurasi analisis sentimen :", akurasi*100, "%")
print()

ulasan_baru = [
    "produk sangat bagus dan memuaskan",
    "barang rusak dan buat kecewa",
    "lumayan bagus tapi pengiriman lambat",
    "lumayan bagus tapi agak lambat",
    "produk tidak berkualitas dan membuat sedih",
]

ulasan_baru_vec = vectorizer.transform(ulasan_baru)
prediksi = model.predict(ulasan_baru_vec)


for i, ulasan in enumerate(ulasan_baru):
  print(f'"{ulasan}"')
  print(f"adalah sentimen : {prediksi[i]}")
  print()
print(vectorizer.get_feature_names_out(ulasan_baru_vec))
```
