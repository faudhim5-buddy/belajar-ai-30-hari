```python
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
import matplotlib.pyplot as plt
import numpy as np

data = np.array([
    [170, 65, 25], [175, 70, 30], [160, 55, 22], [180, 80, 35],
    [165, 60, 28], [172, 68, 32], [158, 52, 20], [178, 75, 40],
    [155, 85, 45], [160, 90, 50], [165, 88, 42], [170, 92, 48],
    [175, 87, 44], [158, 91, 46], [162, 89, 43], [168, 93, 49],
])
label = [
    "aktif", "aktif", "aktif", "aktif",
    "aktif", "aktif", "aktif", "aktif",
    "tidak aktif", "tidak aktif", "tidak aktif", "tidak aktif",
    "tidak aktif", "tidak aktif", "tidak aktif", "tidak aktif",
]

X_latih, X_uji, y_latih, y_uji = train_test_split(
    data, label, test_size=0.2, random_state=42
)

model = DecisionTreeClassifier(max_depth=3, random_state=42)
model.fit(X_latih, y_latih)

akurasi = accuracy_score(y_uji, model.predict(X_uji))
print("Akurasi Decision Tree:", akurasi *100, "%")
print()

plt.figure(figsize=(10,4))
plot_tree(model,
          feature_names=["tinggi","berat","umur"],
          class_names=["aktif","tidak aktif"],
          filled=True,
          rounded=True,
          fontsize=10
          )
plt.title("Visualisasi Decision Tree")
plt.show()
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
import matplotlib.pyplot as plt
import numpy as np

data = np.array([
    [170, 65, 25], [175, 70, 30], [160, 55, 22], [180, 80, 35],
    [165, 60, 28], [172, 68, 32], [158, 52, 20], [178, 75, 40],
    [155, 85, 45], [160, 90, 50], [165, 88, 42], [170, 92, 48],
    [175, 87, 44], [158, 91, 46], [162, 89, 43], [168, 93, 49],
])
label = [
     "aktif", "aktif", "aktif", "aktif",
    "aktif", "aktif", "aktif", "aktif",
    "tidak aktif", "tidak aktif", "tidak aktif", "tidak aktif",
    "tidak aktif", "tidak aktif", "tidak aktif", "tidak aktif",
]

X_latih, X_uji, y_latih, y_uji = train_test_split(
    data, label, test_size=0.2, random_state=42,
)

model_dt = DecisionTreeClassifier(max_depth=3, random_state=42)
model_dt.fit(X_latih, y_latih)
akurasi_dt = accuracy_score(y_uji, model_dt.predict(X_uji))

model_rf = RandomForestClassifier(n_estimators=100, random_state=42)
model_rf.fit(X_latih, y_latih)
akurasi_rf = accuracy_score(y_uji, model_rf.predict(X_uji))

fitur = ["tinggi","berat","umur"]
importance = model_rf.feature_importances_

print("Akurasi Desicion Tree", akurasi_dt *100, "%")
print("Akurasi Random Forest", akurasi_rf *100, "%")

print()
plt.bar(fitur, importance)
plt.title("Features Importance -- Random Forest")
plt.ylabel("Tingkat Kepentingan")
plt.show()

for i, f in enumerate(fitur):
  print(f"{f}: {importance[i]*100:.1f}%")
```
