```python
import numpy as np
import matplotlib.pyplot as plt

data = np.array([
    [160,50], [165,55], [170,60],
    [155,70], [160,75], [156,80],
])

label = ["ideal", "ideal", "ideal", "overweihgt", "overweight", "overweight",]

for i, titik in enumerate(data) :
  warna = "blue" if label[i] =="ideal" else "red"
  plt.scatter(titik[0], titik[1], color=warna)

plt.title("Data Tinggi vs Data Berat")
plt.xlabel("Tinggi")
plt.ylabel("Berat")
plt.legend(label)
plt.show()
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
import numpy as np

data = np.array([
    [160,50], [165,55], [170,60],
    [155,70], [160,75], [156,80],
])

label = ["ideal", "ideal", "ideal", "overweihgt", "overweight", "overweight",]

model = KNeighborsClassifier(n_neighbors=3)
model.fit(data, label)

prediksi = model.predict(data)
akurasi = accuracy_score(label, prediksi)
print("prediksi model :", prediksi)
print("label asli :", label)
print("akurasinya", akurasi*100, "%")

print()
print("Model KKN Sudah dilatih!")
print()

orang_baru = np.array( [[163,58]] )
hasil = model.predict(orang_baru)
print("Prediksi orang dengan tinggi 163cm dan berat 58kg adalah", hasil[0])

orang_lain = np.array([[175,75]])
hasil_2 = model.predict(orang_lain)
print("Prediksi orang lain dengan tinggi 175 berat 75 adalah", hasil_2[0])
```

```python
prediksi = model.predict(data)

for i in range(len(label)):
  cocok = "benar" if prediksi[i] == label[i] else "salah"
  print(f"data {i+1} : Prediksi = {prediksi[i]}, Asli = {label[i]} {cocok}")

print("Akurasi", accuracy_score(label, prediksi) * 100, "%" )
```
