```python
data_mahasiswa = [
    {"nama":"Sonya","nilai":88},
    {"nama":"Yeni","nilai":75},
    {"nama":"Una","nilai":92},
    {"nama":"Yuna","nilai":90},
    {"nama":"Bian","nilai":75},
    {"nama":"Jiya","nilai":97},
    {"nama":"Dhima","nilai":60},
    {"nama":"Buddy","nilai":50},
]

def cek_kelulusan(nilai):
  if nilai >= 90:
    return "A - Cumlaude"
  elif nilai >= 75:
    return "B - Good"
  elif nilai >= 60:
    return "C - Lulus bersyarat"
  else:
    return "D - Ulang semester depan!"

def hitung_rata_rata(data):
  total = 0
  for mahasiswa in data:
    total = total + mahasiswa["nilai"]
    return total / len(data)


print("==== HASIL UJIAN ====")
for mahasiswa in data_mahasiswa:
  status = cek_kelulusan(mahasiswa["nilai"])
  print(mahasiswa["nama"], "dengan Nilai", mahasiswa["nilai"], "dinyatakan", status)

print()
print("Rata-rata Kelas", hitung_rata_rata(data_mahasiswa))
```
